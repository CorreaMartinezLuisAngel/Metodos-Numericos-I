```python
epsilon = 1.0
uno = 1.0

while epsilon + uno > uno:
    epsilon /= 2

print("Epsilon de la máquina:", epsilon)
```

```python
eps = 1.0
while (1.0 + eps) != 1.0:
    eps /= 2.0
print("Epsilon de la máquina:", eps)

x = 1.0
y = eps
suma = x + y
print("Suma de x y y:", suma)
```
