<font color="blue"><h1><b>Iteracion de punto fijo</h1></b></font>

<font face="Georgia" size="3" color="blue">
Objetivo:
</font>
<font face="Georgia" size="3">
Queremos resolver el problema $f(x)=0$.
</font>

<font face="Georgia" size="3" color="blue">
Conceptos y Condiciones:
</font>
<font face="Georgia" size="3">
Tomemos en consideración que un punto fijo es aquel en qué se cumple que $g(p)=p$.<br>
De esta manera queremos transformar nuestro problema $f(x)=0$ en la forma equivalente $g(x)=x$ y así buscamos un valor $p$ tal que $g(p)=p$ y a su vez $f(p)=0$.
</font>

<font face="Georgia" size="3">
Para hallar esta $g(x)$ hay diversas maneras, algunos ejemplos son:<br>
1.$g(x)=x-f(x)$<br>
2.$g(x)=x-2f(x)$<br>
3.$g(x)=x-kf(x)$ con $k≠0$<br>
También podemos despejar directamente x en $f(x)=0$ para encontrar $g(x)$
</font>

<font face="Georgia" size="3" color="blue">
Verificación de las condiciones en [a,b]:<br>
</font>
<font face="Georgia" size="3">
Primero busquemos dos valores $a$ y $b$ tal que evaluados en $f(x)$ sean de signos opuestos, es decir, $f(a)$ sea de signos opuesto a f(b) para garantizar que haya al menos una raíz en ese intervalo.<br>
</font>

<font face="Georgia" size="3" color="blue">
Condiciones para la función propuesta:<br>
</font>
<font face="Georgia" size="3">
Si $g(x)$ es continua en $[a,b]$ y diferenciable en $(a,b)$ por el teorema del valor extremo alcanza un máximo absoluto y un mínimo absoluto dentro del intervalo, ya sea en los extremos o donde $g'(x)=0$.<br>
Después debemos verificar que la imagen de la función no se salga del intervalo, es decir, que $g(x) \in [a,b]$, $\forall x \in [a,b]$.
</font>

<font face="Georgia" size="3">
Ahora usando el teorema del punto fijo:<br>
Sea $g \in C[a,b]$ tal que $g(x) \in [a,b]$, $\forall x\in [a,b]$. Además supongamos que existe $g'$ en $[a,b]$ y una constante $k<1$ con:<br>
$|g'(x)|≤k$, $\forall x \in (a,b)$ <br>
Entonces para cualquier número $P_n$ en $[a,b]$ la sucesión definida por:<br>
$P_n=g(P_{n-1}) , n≥1$<br>
Converge en el único punto fijo $p$ en $[a,b]$
</font>


<font face="Georgia" size="3">
Tomando está parte:<br>
$|g'(x)|≤k<1|$ , $\forall x \in [a,b]$.<br>
Para ello evaluamos $|g'(x)|$ en los extremos del intervalo y asegurándonos de que su valor máximo absoluto sea estrictamente menor que 1.
</font>

</font>
<font face="Georgia" size="3">
Si todas las condiciones anteriores se cumplen entonces para cualquier número inicial $p_0 \in [a,b]$ se tiene la sucesión definida por la fórmula recursiva:<br>
$P_n=g(P_{n-1}) , n≥1$<br>
Converge a un único punto fijo $p \in [a,b]$.
</font>

<font face="Georgia" size="3" color="blue">
Algoritmo:<br>
</font>
<font face="Georgia" size="3">
Finalmente para desarrollar el método tenemos que seleccionar un $P_0 \in [a,b]$ (comúnmente se toma el punto medio) y después calculamos aproximaciones con la $g(x)$ seleccionada, es decir:<br>
$P_1 = g(P_0)$ , $P_2 = g(P_1)$ , $P_3 = g(P_2)$ , ... , $P_n = g(P_{n-1})$.<br>
Detenemos el proceso hasta que se satisface la tolerancia deseada, es decir que el error sea menor a la tolerancia:<br>
$|P_n - P_{n-1} | < TOL$
</font>

<font face="Georgia" size="3" color="blue">
Ejemplo:<br>
</font>
<font face="Georgia" size="3">
Ahora empleado este método resolveremos el ejercicio 1 de la página 38 del libro de Burden:
Muestre que $f(x) = x^3+ 4x^2− 10 = 0$  tiene una solución en $[1,2]$ , determinar una aproximación para la raíz que sea precisa por lo menos dentro  
de $10^{-4}$
</font>


<font face="Georgia" size="3">
Primero veamos si evaluando la función $f(x)=0$ en los extremos del intervalo tienen signos opuestos para garantizar que se encuentre una raíz en el intervalo:
</font>

```python
#Importamos las librerías necesarias
import numpy as np
```

```python
#Definimos la función original
def f(x):
    return (x**3)+4*(x**2)-10

#Definimos los datos que tenemos:
a=1
b=2

#Evaluando en los extremos del intervalo
fa=f(a)
fb=f(b)

print(f"La función evaluada en a es {fa}")
print(f"La función evaluada en b es {fb}")

if np.sign(fa)==np.sign(fb):
    print(f"No se garantiza encontrar una raíz en [{a},{b}]")
```

<font face="Georgia" size="3">
Ahora proponemos una función $g(x)$:<br>
$g(x)=\sqrt \frac {10}{x+4} $<br>
y veamos si cumple las condiciones para llevar a acabo el método.<br>
Primero veamos si es continua en [1,2] y su imagen está dentro de este intervalo.
</font>

```python
#Definimos la función que queremos iterar
def g(x):
    return np.sqrt(10/(x+4))
ga=g(a)
gb=g(b)
print(f"La función evaluada en g(a) es {ga}")
print(f"La función evaluada en g(b) es {gb}")
```

<font face="Georgia" size="3">
Ahora hallemos su derivada y veamos la derivada:<br>
$g'(x)= -\frac{\sqrt {10}}{2 \sqrt {(x+4)^3}}$<br>
Cómo observamos está función es decreciente en $[1,2]$, por lo cual el maximo se encuentra en $2$
</font>

<font face="Georgia" size="3">
Después usando el teorema del punto fijo aseguramos la convergencia:<br>
$|g'(x)|= \frac{\sqrt {10}}{2 \sqrt {(x-4)^3}}$<br>
Y como el maximo de está función está en $2$ entonces:<br>
$|g'(x)|= \frac{\sqrt {10}}{2 \sqrt {(2+4)^3}}=\frac{\sqrt {10}}{2 \sqrt {(6)^3}}=\frac{\sqrt {10}}{2 \sqrt {216}}≈0.10758287$<br>
Por lo tanto $|g'(x)|≤k<1$, y entonces existe un único punto fijo en [1,2].
</font>


<font face="Georgia" size="3">
Ahora procederemos a realizar las iteraciones hasta tener la precisión de $10^{-4}$:
</font>

```python
#Método de Iteración de Punto Fijo
#Definimos los datos
a=1 #Extremo izquierdo
b=2 #Extrwmo derecho
p0=0.5 #Pumto incial para comenzar a iterar
tol=0.0001 #precision
nmax=100 #número máximo de iteraciones
niter=0 #número de Iteración

#Evualuando la función en los extremos
ga=g(a)
gb=g(b)

#Evaluando en el primer p dado
gp0=g(p0)
print(f"La función evaluada en g(p0) es: {gp0}")

print("-" * 50) #Linea de separación

#Impresión de la Iteracion 0
print("iter\tp\t\tg(p)\t\terror")
print("{0}\t{1:6.6f}\t{2:6.6f}\t{3:6.6f}".format(niter, p0, gp0, 0.0))

#Error incial para arrancar el ciclo iterativo
error = tol + 1

#Ciclo iterativo
while error>tol and niter<nmax:
    p=g(p0)
    error=abs(p-p0)
    gp=g(p)
    niter+=1
    print("{0}\t{1:.6f}\t{2:.6f}\t{3:.6f}".format(niter, p, gp, error))
    p0=p
    if error <= tol:
         break
print("-" * 50) #Linea de separación

#Solucion
if niter>=nmax:
        print(f"El método fracasó después de {niter} iteraciones.")
else:
        print(f"La raíz de la función dada en el intervalo [{a:6.4f},{b:6.4f}] es {gp:.6f}".format(a, b, gp))
```

```python
#Veamos una gráfica de la función que iteramos
import numpy as np
import matplotlib.pyplot as plt

#Impresión de la funcion
xx=np.linspace(-1,5,100)
plt.plot(xx,g(xx))

#Impresión de los ejes
plt.axhline(0, color='black')  # Eje X
plt.axvline(0, color='black')  # Eje Y

#Impresion de los puntos
plt.plot(a,0,'ro') #Punto incial a
plt.plot(b,0,'ro') #Punto incial b
plt.plot(p,0,'go') #Aproximacion obtenida
plt.plot(0,gp,'go')

#Rotulos
plt.text(a,0.1, 'a', color='red', fontsize=12, ha='center')
plt.text(b,0.1, 'b', color='red', fontsize=12, ha='center')
plt.text(p,0.1, 'p', color='green', fontsize=12, ha='center')
plt.text(-0.3,gp,'g(p)',color='green', fontsize=12,ha='center')

#Rectas:
plt.plot([p,p],[0,gp],'g--')
plt.plot([0,p],[gp,gp],'g--')

#Cuadricula
plt.grid()
```
