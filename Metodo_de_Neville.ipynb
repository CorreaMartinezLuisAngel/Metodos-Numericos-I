<font color="gree"><h1><b>Método de Neville</b></h1></font>

<font face="Georgia" size="3" color="gree">
Objetivo:
</font>
<font face="Georgia" size="3">
El objetivo de este método es aproximar el valor de una función desconocida en un punto $x$, conociendo únicamente un conjunto de nodos
</font>

<font face="Georgia" size="3" color="gree">
¿En qué consiste?:
</font>
<font face="Georgia" size="3">
El método de Neville es un algoritmo de interpolación polinómica desarrollado por el matemático británico Eric Harold Neville.

El algoritmo busca calcular el valor de un polinomio interpolador de grado $n$ evaluado en un punto específico construyendo los resultados de forma iterativa a partir de polinomios de menor grado, el método construye una tabla triangular donde cada entrada se calcula a partir de los valores adyacentes de la columna anterior mediante una relación de recurrencia que está dada por:
</font>

$$
Q_{i,j}(x)=\frac{(x-x_{i-j})Q_{i,j-1}-(x-x_i)Q_{i-1,j-1}}{x_i-x_{i-j}}
$$

<font face="Georgia" size="3">
El siguiente código es una representación de método usando como ejemplo el ejercicio 2 de la página 89 del libro de Burden:
</font>

```python
#Datos
#Datos
x=1.5 #Valor que se desea aproximar
xi=[1.0, 1.3, 1.6, 1.9, 2.2] #Nodos
fi=[0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623] #Funcion evaluada en los nodos
n=len(xi)-1 #Cantidad de nodos
```

```python
#Metodo de Neville
#Se genera una matriz de ceros
Q=[[0.0 for j in range(n+1)] for i in range(n+1)]

#Se añaden los polinomios de grado cero (fi)
for i in range(n + 1):
    Q[i][0] = fi[i]

#Formula de Neville
for i in range(1, n + 1):
    for j in range(1, i + 1):
        Q[i][j] = ((x - xi[i - j]) * Q[i][j - 1] - (x - xi[i]) * Q[i - 1][j - 1]) / (xi[i] - xi[i - j])
```

```python
#Tabla de Resultados
print("Tabla de Neville:")
for i in range(n+1):
    for j in range(i+1):
        print(f"{Q[i][j]:12.6f}",end="")
    print()

print(f"\nEl valor interpolado P({x}) = {Q[n][n]:.6f}")
```

<font face="Georgia" size="3">
El método de Neville e suba gran herramienta de interpolación polinomica, lo cual le da una gran ventaja ya que los polinomios son faciles de manejar, además de que si lo deseamos podemos calcular directamente la aproximación o calcular el polinomio.
También gracias a su tabla se pueden conseguir varias aproximaciones combinando los polinomios de grado menor, lo que facilita el trabajo computacional.
Algo a tomar en cuenta es que si queremos tener una buena aproximación el valor que se desea aproximar debe estar entre los nodos seleccionados ya que si este se encuentra fuera de estos nodos se estaría haciendo una extrapolación por lo que la aproximación sería más imprecisa.
